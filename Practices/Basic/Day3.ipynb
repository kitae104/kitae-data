```python
import pandas as pd
import numpy as np

data = {
    '학생ID': [101, 102, 103, 104, 105, 106, 107, 108],
    '이름': ['김민수', '이지은', '박준호', '최서연',
            '정하늘', '강지훈', '윤서진', '한도윤'],
    'Python': [85, 92, np.nan, 78, 95, 70, np.nan, 88],
    'SQL': [90, np.nan, 75, 88, 92, np.nan, 85, 80]
}

df = pd.DataFrame(data)
df2 = df.copy()
df
```

#### Python과 SQL 과목에 결측치가 각각 몇 개 있는지 확인하세요

```python
df2.isnull().sum()
```

#### 각 과목의 결측치를 해당 과목의 **중앙값(Median)** 으로 대체하세요.

```python
cols = ['Python', 'SQL']
for col in cols:
    median_value = df2[col].median()
    df2[col] = df2[col].fillna(median_value)
    
df2
```

#### 결측치를 처리한 후 학생별 평균 점수를 계산하세요.

```python
df2['평균'] = df2[['Python', 'SQL']].mean(axis=1)
```

```python
result = df2.sort_values(by='평균', ascending=False )
result
```
