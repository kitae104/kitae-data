```python
import pandas as pd

data = {
    '상품명': ['노트북', '마우스', '키보드', '책상', '의자', '모니터'],
    '카테고리': ['전자제품', '전자제품', '전자제품', '가구', '가구', '전자제품'],
    '가격': [1200000, 30000, 80000, 250000, 150000, 350000],
    '판매수량': [5, 20, 12, 4, 8, 6]
}

df = pd.DataFrame(data)

df
```

```python
df['매출액'] = df['가격'] * df['판매수량']
df
```

```python
df_filtered = df[(df['매출액'] > 1000000) & (df['카테고리'] == '전자제품')]
df_filtered.sort_values(by='매출액', ascending=False)
df_filtered
```
