```python
import pandas as pd

df = pd.DataFrame({
    '이름': ['김철수', '이영희', '박민수',
            '최지은', '정하늘', '강민호'],
    '과목': ['Python', 'Python', 'Java',
            'Java', 'Python', 'Java'],
    '점수': [85, 92, 78, 88, 76, 95]
})
df
```

### 오늘의 문제
과목별 평균 점수를 구하고, 평균이 높은 과목부터 내림차순으로 출력하세요.
가능하면 다음 세 가지 조건도 만족하도록 작성해 보세요.
- groupby()를 사용할 것
- 평균 점수의 열 이름을 평균점수로 지정할 것
- 결과를 DataFrame 형태로 출력할 것

```python
result = df.groupby('과목')['점수'].mean().reset_index(name="평균점수").sort_values("평균점수", ascending=False)
result
```

### 같은 데이터에서 80점 이상인 학생만 선택한 다음 과목별 평균 점수를 구해 보세요.

```python
df[df['점수'] >= 80].groupby('과목')['점수'].mean().reset_index(name="평균점수").sort_values("평균점수", ascending=False)
```
