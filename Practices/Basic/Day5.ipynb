```python
import pandas as pd

products = pd.DataFrame({
    '제품ID': ['P001', 'P002', 'P003', 'P004', 'P005'],
    '제품명': ['노트북', '마우스', '키보드', '모니터', '웹캠'],
    '가격': [1200000, 30000, 80000, 350000, 90000]
})

sales = pd.DataFrame({
    '주문ID': ['O101', 'O102', 'O103', 'O104',
             'O105', 'O106', 'O107'],
    '제품ID': ['P001', 'P002', 'P001', 'P003',
             'P004', 'P002', 'P999'],
    '판매수량': [2, 5, 1, 3, 2, 10, 4]
})
```

```python
products.head()
```

```python
sales.head()
```

1. 제품ID를 기준으로 두 DataFrame을 결합하세요.

```python
df = pd.merge(sales, products, on='제품ID', how='left', validate='many_to_one', indicator=True)
df
```

2. 제품 정보가 존재하지 않는 주문을 찾으세요.

```python
df[df['제품명'].isnull()]
```

3. 정상 주문의 매출액을 계산하세요.

```python
df_order = df[df['제품명'].notnull()]
df_order['매출액'] = df_order['판매수량'] * df_order['가격']
df_order
```

4. 제품별 총판매수량과 총매출액을 계산하세요.

```python
df_order.groupby('제품명')[['판매수량', '매출액']].sum()
```

5. 매출액이 높은 순서대로 정렬하세요.

```python
df_order.groupby('제품명')[['판매수량', '매출액']].sum().sort_values(by='매출액', ascending=False)
```
