### 준비 · 1~4단계 코드 실행 (df_clean 만들기)

앞 단계에서 만든 정리된 데이터 `df_clean`을 다시 만듭니다. `result` 열은 '합격'/'불합격' 문자열입니다.

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams['font.family'] = 'Malgun Gothic'  # 한글 표시 (Mac: 'AppleGothic')
plt.rcParams['axes.unicode_minus'] = False       # 음수(-) 기호 깨짐 방지

df = pd.read_csv('반도체_공정_샘플.csv')

# 3단계: 숫자 결측값은 평균으로 채우고 중복 제거
numeric_cols = ['온도_섭씨', '압력_Pa', '가스유량_slm', '전력_W',
                '진공도_mTorr', '두께_nm', '습도_pct', '진동_mm_s',
                '처리시간_sec', '냉각수온도_섭씨']
df_clean = df.copy()
for col in numeric_cols:
    df_clean[col] = df_clean[col].fillna(df_clean[col].mean())
df_clean = df_clean.drop_duplicates().reset_index(drop=True)

# 4단계: 합격여부 1/-1 → '합격'/'불합격'
df_clean['result'] = df_clean['합격여부'].map({1: '합격', -1: '불합격'})

print(f"준비 완료: df_clean {df_clean.shape[0]}행 × {df_clean.shape[1]}열")
````

````python
vib_90 = df_clean['진동_mm_s'].quantile(0.9)
high_vib = df_clean[df_clean['진동_mm_s'] >= vib_90]

all_fail = (df_clean['합격여부'] == -1).mean() * 100
vib_fail = (high_vib['합격여부'] == -1).mean() * 100

print(f"진동 상위 10% 기준: {vib_90:.3f} mm/s 이상 ({len(high_vib)}건)")
print(f"전체 불합격 비율: {all_fail:.1f}%")
print(f"진동 상위 10% 불합격 비율: {vib_fail:.1f}%")
print('비율 차이는 원인 확정이 아니라 추가 점검이 필요한 단서입니다.')
````
