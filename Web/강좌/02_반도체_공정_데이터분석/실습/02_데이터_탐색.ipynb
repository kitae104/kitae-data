### 1 CSV 파일 읽어서 DataFrame 만들기
- 원본 코드: `02_read_csv.py`
- 설명: CSV 파일을 pandas DataFrame으로 읽어 분석 준비를 합니다.

### 준비 · 라이브러리 불러오기 (1단계와 동일)

```python
# 📦 분석에 사용할 라이브러리 불러오기
# 'as pd'는 앞으로 pandas를 pd라는 짧은 이름으로 부르겠다는 약속
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 그래프 한글 설정 (Mac은 'AppleGothic')
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False   # 음수(-) 기호 깨짐 방지
```

```python
# 📂 CSV 파일을 읽어서 df라는 이름의 DataFrame으로 저장
df = pd.read_csv('반도체_공정_샘플.csv')

# 데이터 크기 확인: (행의 수, 열의 수)
print(f"데이터 크기: {df.shape[0]}행 × {df.shape[1]}열")
```

### 2 데이터를 처음 받으면 꼭 해보는 5가지
- 원본 코드: `03_explore.py`
- 설명: head, info, describe 등으로 데이터의 첫인상과 구조를 확인합니다.

```python
# ① 처음 5줄 미리보기
df.head()

# ② 데이터 크기
print("크기:", df.shape)

# ③ 전체 구조 요약 (열 이름·타입·결측 수)
df.info()

# ④ 숫자 열의 기술 통계
df.describe().round(2)   # 보기 좋게 소수점 2자리 반올림
```

### 3 원하는 "열(컬럼)"만 골라 보기
- 원본 코드: `04_select_columns.py`
- 설명: 필요한 행과 열만 골라 보며 DataFrame 선택 문법을 연습합니다.

```python
## 한 개 열만 선택 → 결과는 Series (열 하나짜리)
df['온도_섭씨']            # 온도_섭씨 한 줄만 200개

## 여러 열 선택 → 결과는 DataFrame (대괄호 두 겹!)
df[['온도_섭씨', '압력_Pa', '합격여부']]

## 한 개 열의 통계 계산
print(f"평균 온도: {df['온도_섭씨'].mean():.2f}°C")
print(f"최고 온도: {df['온도_섭씨'].max():.2f}°C")
```

### 4 원하는 "행"만 골라 보기 — loc vs iloc
- 원본 코드: `05_loc_iloc.py`
- 설명: loc/iloc을 사용해 라벨 기준과 위치 기준 선택의 차이를 확인합니다.

```python
## iloc - 위치(번호)로 선택. 0부터 시작!
df.iloc[0]          # 첫 번째 행 (가장 위 한 줄)
df.iloc[1:5]       # 1~4번 행 (5번 미포함!)

## loc - 인덱스 이름(라벨)으로 선택
df.loc[0]              # 인덱스가 0인 행
df.loc[0:4, ['온도_섭씨', '합격여부']]  # 행+열 동시 지정
```

### 5 카테고리 값 세기 — value_counts()
- 원본 코드: `06_value_counts.py`
- 설명: 범주형 값이 몇 번씩 등장하는지 세어 데이터 분포를 파악합니다.

```python
# 합격여부 컬럼의 값별 개수
print(df['합격여부'].value_counts())

# 비율(%)로 보고 싶다면 normalize=True
print(df['합격여부'].value_counts(normalize=True) * 100)
```
