### 2 결측값 찾기
- 설명: 결측값을 찾고 삭제 또는 대체 방식으로 정리합니다.

### 준비 · 1~2단계 코드 실행 (라이브러리 + CSV 읽기)

```python
# 📦 분석에 사용할 라이브러리 불러오기
# 'as pd'는 앞으로 pandas를 pd라는 짧은 이름으로 부르겠다는 약속
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 그래프 한글 설정 (Mac은 'AppleGothic')
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False   # 음수(-) 기호 깨짐 방지

df = pd.read_csv('반도체_공정_샘플.csv')
print(f"데이터 크기: {df.shape[0]}행 × {df.shape[1]}열")
```

```python
# ① 열별 결측값 개수 (True를 1로 세어줌)
print(df.isnull().sum())
```

```python
# ② 전체 결측값 개수 (모든 열을 합산)
print(f"전체 결측값 개수: {df.isnull().sum().sum()}개")
```

### 3 결측값 처리하는 3가지 방법
- 설명: 결측값을 찾고 삭제 또는 대체 방식으로 정리합니다.

```python
# 원본 보호를 위해 복사본을 만들어 작업
df_filled = df.copy()

# 숫자 컬럼만 골라서 평균으로 채우기
numeric_cols = ['온도_섭씨', '압력_Pa', '가스유량_slm', '전력_W',
                '진공도_mTorr', '두께_nm', '습도_pct', '진동_mm_s',
                '처리시간_sec', '냉각수온도_섭씨']

for col in numeric_cols:
    mean_val = df_filled[col].mean()
    df_filled[col] = df_filled[col].fillna(mean_val)

# 처리 후 결측 확인
print(f"처리 후 남은 결측값: {df_filled.isnull().sum().sum()}개")
```

### 4 중복된 행 찾고 제거하기
- 설명: 중복 행을 확인하고 제거한 뒤 인덱스를 정리합니다.

```python
# ① 중복된 행 개수 확인
print(f"중복 행 수: {df_filled.duplicated().sum()}개")
```

```python
# ② 중복 제거
df_clean = df_filled.drop_duplicates()

# ③ 인덱스 다시 매기기 (중복 제거 후 번호가 띄엄띄엄 됨)
df_clean = df_clean.reset_index(drop=True)

print(f"정리 전: {len(df_filled)}행 → 정리 후: {len(df_clean)}행")
```

## 마무리

- 원본 `df`는 그대로 두고 복사본 `df_filled`에서 결측값을 채웠습니다.
- 중복 제거 후 인덱스를 다시 매긴 `df_clean`이 4단계 EDA의 출발점입니다.
