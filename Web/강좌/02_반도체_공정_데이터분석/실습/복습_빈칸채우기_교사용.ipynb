## 3단계 · 결측값 & 중복 처리

### Q11. 결측값 개수 세기

열별 결측값 개수와 전체 결측값 개수를 봅니다.

**정답:** `isnull` · `sum`

```python
print(df.isnull().sum())
print(f"전체 결측값 개수: {df.isnull().sum().sum()}개")
```

### Q12. 결측값을 평균으로 채우기

원본은 두고 복사본의 숫자 열 결측값을 각 열의 평균으로 채웁니다.

**정답:** `copy` · `fillna` · `mean`

```python
df_filled = df.copy()
numeric_cols = ['온도_섭씨', '압력_Pa', '가스유량_slm', '전력_W',
                '진공도_mTorr', '두께_nm', '습도_pct', '진동_mm_s',
                '처리시간_sec', '냉각수온도_섭씨']

for col in numeric_cols:
    df_filled[col] = df_filled[col].fillna(df_filled[col].mean())

print(f"처리 후 남은 결측값: {df_filled.isnull().sum().sum()}개")
```

### Q13. 중복 제거하고 인덱스 다시 매기기

중복 행 수를 확인하고, 제거한 뒤 인덱스를 0부터 다시 매깁니다.

**정답:** `duplicated` · `drop_duplicates` · `reset_index`

```python
print(f"중복 행 수: {df_filled.duplicated().sum()}개")
df_clean = df_filled.drop_duplicates().reset_index(drop=True)
print(f"정리 전: {len(df_filled)}행 → 정리 후: {len(df_clean)}행")
```

## 4단계 · 탐색적 데이터 분석 (EDA)

### Q14. 값을 바꿔 새 열 만들기

합격여부 1/-1을 '합격'/'불합격'으로 바꾼 result 열을 만듭니다.

**정답:** `map`

```python
df_clean['result'] = df_clean['합격여부'].map({1: '합격', -1: '불합격'})
df_clean[['합격여부', 'result']].head()
```

### Q15. 불합격 데이터만 고르기

합격여부가 -1인 행만 골라 개수를 셉니다.

**정답:** `==`

```python
fail_df = df_clean[df_clean['합격여부'] == -1]
print(f"불합격 건수: {len(fail_df)}건")
```

### Q16. 여러 조건 함께 쓰기

온도가 305도를 넘고 두께가 103nm를 넘는 공정을 고릅니다.

**정답:** `&`

```python
risk_df = df_clean[(df_clean['온도_섭씨'] > 305) & (df_clean['두께_nm'] > 103)]
print(f"고온+두꺼움 공정: {len(risk_df)}건")
```

### Q17. 합격/불합격별 평균 비교하기

result별 숫자 열 평균을 구하고, 보기 편하게 행과 열을 뒤집어 출력합니다.

**정답:** `groupby` · `T`

```python
group_mean = df_clean.groupby('result')[numeric_cols].mean().round(2)
print(group_mean.T)
```

### Q18. 히스토그램 그리기

온도 값이 어떻게 퍼져 있는지 봅니다.

**정답:** `hist`

```python
plt.figure(figsize=(8, 4))
plt.hist(df_clean['온도_섭씨'], bins=20, color='steelblue', edgecolor='white')
plt.title('온도 분포')
plt.xlabel('온도 (°C)')
plt.ylabel('빈도 (개수)')
plt.show()
```

### Q19. 박스플롯으로 합격/불합격 비교하기

seaborn으로 판정별 온도 분포를 비교합니다.

**정답:** `boxplot`

```python
plt.figure(figsize=(7, 5))
sns.boxplot(data=df_clean, x='result', y='온도_섭씨',
            hue='result', legend=False,
            palette=['#52B788', '#E76F51'])
plt.title('판정별 온도 분포 비교')
plt.show()
```

### Q20. 상관관계 히트맵 그리기

숫자 열과 합격여부 사이의 상관관계를 계산하고 색으로 봅니다.

**정답:** `corr` · `heatmap`

```python
corr_matrix = df_clean[numeric_cols + ['합격여부']].corr()

plt.figure(figsize=(11, 9))
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='RdYlGn', center=0, linewidths=0.5)
plt.title('변수 간 상관관계')
plt.show()
```

## 5 · 보강 실습

### Q21. 결측률(%) 구하기

원본 df의 열별 결측 비율을 %로 보고, 큰 순서로 5개만 봅니다.

**정답:** `mean`

```python
missing_rate = df.isna().mean() * 100
print(missing_rate.sort_values(ascending=False).head().round(2))
```

### Q22. IQR로 이상값 후보 찾기

온도의 Q1(25%), Q3(75%)로 일반 범위를 구하고, 범위를 벗어난 행을 셉니다.

**정답:** `quantile` · `0.75` · `|`

```python
q1 = df_clean['온도_섭씨'].quantile(0.25)
q3 = df_clean['온도_섭씨'].quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

outliers = df_clean[(df_clean['온도_섭씨'] < lower) | (df_clean['온도_섭씨'] > upper)]
print(f"온도 일반 범위: {lower:.2f} ~ {upper:.2f} °C")
print(f"이상값 후보: {len(outliers)}건")
```

### Q23. 평균 차이가 큰 변수 찾기

불합격 평균과 합격 평균의 차이를 절댓값으로 구해 큰 순서로 봅니다.

**정답:** `abs`

```python
result_means = df_clean.groupby('result')[numeric_cols].mean()
mean_gap = (result_means.loc['불합격'] - result_means.loc['합격']).abs()
print(mean_gap.sort_values(ascending=False).head().round(3))
```
