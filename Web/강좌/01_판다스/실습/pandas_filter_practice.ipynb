# Pandas 필터링 연습 예제

학생 성적 데이터를 이용해 Pandas의 다양한 필터링 방법을 연습합니다.

주요 내용:
- 단일 조건 필터
- AND / OR / NOT 조건
- `isin()`, `between()`
- `loc`
- 문자열 필터
- 열 간 비교
- 최고값/평균값 기반 필터
- 정렬과 상위 데이터 추출
- `query()` 사용


```python
import pandas as pd

data = {
    '이름': ['김민준', '이서연', '박지호', '최예린', '정도윤'],
    '국어': [85.0, 92.0, 78.0, 85.8, 88.0],
    '수학': [92.0, 78.0, 82.5, 88.0, 72.0],
    '영어': [88.0, 86.5, 82.0, 91.0, 85.0],
    '총점': [265.0, 256.5, 242.5, 264.8, 245.0],
    '평균': [88.3, 85.5, 80.8, 88.3, 81.7],
    '학년': [1, 1, 2, 2, 3],
    '등급': ['B', 'B', 'B', 'B', 'B']
}

df = pd.DataFrame(data)
df
```

## 1. 국어 점수가 85점 이상인 학생


```python
df[df['국어'] >= 85]
```

## 2. 수학 점수가 80점 미만인 학생


```python
df[df['수학'] < 80]
```

## 3. 평균이 85점 이상인 학생


```python
df[df['평균'] >= 85]
```

## 4. 1학년 학생만 조회


```python
df[df['학년'] == 1]
```

## 5. 1학년이 아닌 학생


```python
df[df['학년'] != 1]
```

## 6. 국어 85점 이상이면서 수학 85점 이상


```python
df[(df['국어'] >= 85) & (df['수학'] >= 85)]
```

## 7. 국어가 90점 이상이거나 영어가 90점 이상


```python
df[(df['국어'] >= 90) | (df['영어'] >= 90)]
```

## 8. 국어, 수학, 영어가 모두 80점 이상


```python
df[(df['국어'] >= 80) & (df['수학'] >= 80) & (df['영어'] >= 80)]
```

## 9. 평균이 80점 이상 85점 미만


```python
df[(df['평균'] >= 80) & (df['평균'] < 85)]
```

## 10. between()으로 평균 80~85점 조회


```python
df[df['평균'].between(80, 85)]
```

## 11. 1학년 또는 2학년 학생


```python
df[(df['학년'] == 1) | (df['학년'] == 2)]
```

## 12. isin()으로 1학년 또는 2학년 학생 조회


```python
df[df['학년'].isin([1, 2])]
```

## 13. 1, 2학년이 아닌 학생


```python
df[~df['학년'].isin([1, 2])]
```

## 14. 평균 85점 이상 학생의 이름과 평균


```python
df[df['평균'] >= 85][['이름', '평균']]
```

## 15. loc를 이용해 평균 85점 이상 학생의 이름과 평균 조회


```python
df.loc[df['평균'] >= 85, ['이름', '평균']]
```

## 16. 2학년이면서 평균 85점 이상


```python
df[(df['학년'] == 2) & (df['평균'] >= 85)]
```

## 17. 2학년이면서 평균 85점 이상인 학생의 이름과 평균


```python
df.loc[(df['학년'] == 2) & (df['평균'] >= 85), ['이름', '평균']]
```

## 18. 이름이 김민준인 학생


```python
df[df['이름'] == '김민준']
```

## 19. 여러 학생 이름 검색


```python
df[df['이름'].isin(['김민준', '박지호'])]
```

## 20. 성이 '김'인 학생


```python
df[df['이름'].str.startswith('김')]
```

## 21. 이름이 '연'으로 끝나는 학생


```python
df[df['이름'].str.endswith('연')]
```

## 22. 이름에 '지'가 들어가는 학생


```python
df[df['이름'].str.contains('지')]
```

## 23. 국어 점수가 자신의 평균보다 높은 학생


```python
df[df['국어'] > df['평균']]
```

## 24. 국어 점수가 수학 점수보다 높은 학생


```python
df[df['국어'] > df['수학']]
```

## 25. 수학 점수가 국어 점수보다 높은 학생


```python
df[df['수학'] > df['국어']]
```

## 26. 국어 최고점 학생


```python
df[df['국어'] == df['국어'].max()]
```

## 27. 수학 최고점 학생


```python
df[df['수학'] == df['수학'].max()]
```

## 28. 영어 최고점 학생


```python
df[df['영어'] == df['영어'].max()]
```

## 29. 전체 학생 평균 점수 확인


```python
df['평균'].mean()
```

## 30. 전체 학생의 평균 점수보다 높은 학생


```python
df[df['평균'] >= df['평균'].mean()]
```

## 31. 평균 상위 3명


```python
df.sort_values('평균', ascending=False).head(3)
```

## 32. 평균 상위 3명의 이름과 평균


```python
df.sort_values('평균', ascending=False)[['이름', '평균']].head(3)
```

## 33. query()로 국어 85점 이상 조회


```python
df.query('국어 >= 85')
```

## 34. query()로 국어 85점 이상 AND 수학 80점 이상


```python
df.query('국어 >= 85 and 수학 >= 80')
```

## 35. query()로 국어 90점 이상 OR 영어 90점 이상


```python
df.query('국어 >= 90 or 영어 >= 90')
```

## 36. query()로 2학년이면서 평균 85점 이상


```python
df.query('학년 == 2 and 평균 >= 85')
```

## 37. 한 과목이라도 80점 미만인 학생


```python
df[(df['국어'] < 80) | (df['수학'] < 80) | (df['영어'] < 80)]
```

## 38. 모든 과목이 85점 이상인 학생


```python
df[(df['국어'] >= 85) & (df['수학'] >= 85) & (df['영어'] >= 85)]
```

## 39. 영어 85점 이상이면서 총점 250점 이상


```python
df[(df['영어'] >= 85) & (df['총점'] >= 250)]
```
