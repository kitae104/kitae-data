### 응용 2 · 공정별 합격률을 막대그래프로 비교하기

**목표:** 공정(식각·증착·세정)마다 합격률(%)을 계산하고 막대그래프로 비교합니다.

````python
# TODO 1: 합격여부가 1이면 True가 되는 '합격' 열을 만드세요.
# TODO 2: 공정별로 '합격' 열의 평균을 구하고 100을 곱해 합격률(%)을 만드세요.
# TODO 3: 합격률을 막대그래프로 그리고 제목과 축 이름을 붙이세요.
````

<details>
<summary><strong>힌트 보기</strong></summary>

True는 1, False는 0으로 계산되므로 True/False 열의 **평균이 곧 비율**입니다. `groupby('공정')['합격'].mean()`에 100을 곱하고, `plt.bar(결과.index, 결과.values)`로 그립니다.

<summary><strong>AI에게 힌트만 받기</strong></summary>


</details>

**정답 예제:** 먼저 직접 시도한 뒤 아래 셀을 실행하세요.

````python
process_df['합격'] = process_df['합격여부'] == 1
pass_rate = (process_df.groupby('공정')['합격'].mean() * 100).round(1)
print(pass_rate)

plt.bar(pass_rate.index, pass_rate.values, color='steelblue')
plt.title('공정별 합격률')
plt.xlabel('공정')
plt.ylabel('합격률 (%)')
plt.ylim(0, 100)
plt.show()
````
