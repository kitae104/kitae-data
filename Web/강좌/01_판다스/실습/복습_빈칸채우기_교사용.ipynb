## 2차시 · EDA

결측값·중복을 정리하고, 조건 필터와 groupby로 데이터를 탐색합니다.

### 준비 · 결측값과 중복이 있는 데이터

빈칸이 없는 셀입니다. `None`은 값이 없다는 뜻이고, 마지막 행은 일부러 중복으로 넣었습니다.

```python
df_na = pd.DataFrame({
    '이름': ['김민준', '이서연', '박지호', '최예린', '최예린'],
    '국어': [85, 92, None, 95, 95],
    '수학': [92, None, 95, 88, 88],
    '영어': [88, 95, 82, None, None]
})
print(df_na)
```

### Q11. 열별 결측값 개수 세기

각 열에 빈 값이 몇 개인지 셉니다.

**정답:** `isnull` · `sum`

```python
print(df_na.isnull().sum())
```

### Q12. 결측값이 있는 행 삭제하기

빈 값이 하나라도 있는 행을 지운 결과를 봅니다.

**정답:** `dropna`

```python
print(df_na.dropna())
```

### Q13. 결측값을 평균으로 채우기

원본은 두고 복사본에서 국어의 빈 값을 국어 평균으로 채웁니다.

**정답:** `fillna` · `mean`

```python
filled = df_na.copy()
filled['국어'] = filled['국어'].fillna(filled['국어'].mean())
print(filled)
```

### Q14. 중복 행 개수 세기

완전히 같은 행이 몇 개인지 셉니다.

**정답:** `duplicated`

```python
print(df_na.duplicated().sum())
```

### Q15. 중복 제거하고 인덱스 다시 매기기

중복을 지운 뒤 인덱스를 0부터 다시 붙입니다.

**정답:** `drop_duplicates` · `reset_index`

```python
clean = df_na.drop_duplicates().reset_index(drop=True)
print(clean)
```

### Q16. 새 열 만들기

세 과목의 총점과 평균 열을 추가합니다.

**정답:** `3`

```python
df['총점'] = df['국어'] + df['수학'] + df['영어']
df['평균'] = (df['총점'] / 3).round(1)
df
```

### Q17. 조건으로 행 고르기

평균이 85점 이상인 학생만 고릅니다.

**정답:** `>=`

```python
df[df['평균'] >= 85]
```

### Q18. 여러 조건 함께 쓰기

1학년이면서 평균 85점 이상인 학생을 고릅니다.

**정답:** `&`

```python
df[(df['학년'] == 1) & (df['평균'] >= 85)]
```

### Q19. 평균이 높은 순서로 정렬하기

평균 열 기준 내림차순으로 정렬합니다.

**정답:** `sort_values` · `False`

```python
df.sort_values('평균', ascending=False)
```

### Q20. 그룹별 평균 구하기

학년별 국어·수학·영어 평균을 구합니다.

**정답:** `groupby` · `mean`

```python
df.groupby('학년')[['국어', '수학', '영어']].mean().round(1)
```

## 4차시 · 공정 데이터 실습

배운 기능을 반도체 공정 데이터에 이어서 사용합니다.

### 준비 · 공정 이력 데이터

빈칸이 없는 셀입니다. `합격여부`는 1 = 합격, -1 = 불합격이고, 마지막 행은 중복입니다.

```python
process_df = pd.DataFrame({
    'lot_id': ['L001', 'L002', 'L003', 'L004', 'L005', 'L005'],
    '공정': ['식각', '증착', '식각', '세정', '증착', '증착'],
    '온도_섭씨': [298.5, 302.1, None, 299.8, 305.2, 305.2],
    '압력_Pa': [101.2, 99.8, 103.1, 100.5, 98.9, 98.9],
    '두께_nm': [100.4, 102.3, 98.7, 101.1, 104.8, 104.8],
    '합격여부': [1, 1, -1, 1, -1, -1]
})
print(process_df)
```

### Q26. 원본을 복사해서 정리하기

원본은 두고 복사본에서 온도 결측값을 중앙값으로 채우고 중복을 지웁니다.

**정답:** `copy` · `median`

```python
work = process_df.copy()
work['온도_섭씨'] = work['온도_섭씨'].fillna(work['온도_섭씨'].median())
work = work.drop_duplicates().reset_index(drop=True)
print(work)
```

### Q27. 값을 바꿔 새 열 만들기

합격여부 1/-1을 '합격'/'불합격' 글자로 바꾼 판정 열을 만듭니다.

**정답:** `map`

```python
work['판정'] = work['합격여부'].map({1: '합격', -1: '불합격'})
print(work[['lot_id', '합격여부', '판정']])
```

### Q28. 기준값과의 차이 계산하기

두께가 기준 100nm에서 얼마나 벗어났는지 절댓값으로 계산합니다.

**정답:** `abs`

```python
work['두께편차_nm'] = (work['두께_nm'] - 100).abs()
print(work[['lot_id', '두께_nm', '두께편차_nm']])
```

### Q29. 공정별 요약표 만들기

공정별 평균 온도와 평균 두께를 구합니다.

**정답:** `공정`

```python
work.groupby('공정')[['온도_섭씨', '두께_nm']].mean().round(2)
```
