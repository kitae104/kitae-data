# Pandas `loc` / `iloc` 연습

학생 성적 데이터를 이용하여 Pandas의 `loc`과 `iloc`을 단계별로 연습합니다.

## 학습 목표
- `loc`과 `iloc`의 차이 이해
- 행/열 선택
- 범위 선택
- 특정 셀 선택
- 여러 행/열 선택
- 조건식과 `loc` 결합
- 값 수정
- 슬라이싱 차이 이해

```python
import pandas as pd

data = {
    '이름': ['김민준', '이서연', '박지호', '최예린', '정도윤'],
    '국어': [85.0, 92.0, 78.0, 85.8, 88.0],
    '수학': [92.0, 78.0, 82.5, 88.0, 72.0],
    '영어': [88.0, 86.5, 82.0, 91.0, 85.0],
    '총점': [265.0, 256.5, 242.5, 264.8, 245.0],
    '평균': [88.3, 85.5, 80.8, 88.3, 81.7],
    '학년': [1, 1, 2, 2, 3],
    '등급': ['B', 'B', 'B', 'B', 'B']
}

df = pd.DataFrame(data)
df
```

## 1. `loc`과 `iloc`의 기본 차이

- `loc` : **행/열의 이름(label)** 을 기준으로 선택
- `iloc` : **행/열의 위치(position)** 를 기준으로 선택

## 2. 첫 번째 행 선택 - `iloc`

```python
df.iloc[0]
```

## 3. 인덱스 라벨이 0인 행 선택 - `loc`

```python
df.loc[0]
```

## 4. 세 번째 행 선택 - `iloc`

```python
df.iloc[2]
```

## 5. 인덱스 라벨이 2인 행 선택 - `loc`

```python
df.loc[2]
```

## 6. 첫 번째 열 선택 - `iloc`

```python
df.iloc[:, 0]
```

## 7. '이름' 열 선택 - `loc`

```python
df.loc[:, '이름']
```

## 8. 세 번째 열 선택 - `iloc`

```python
df.iloc[:, 2]
```

## 9. '수학' 열 선택 - `loc`

```python
df.loc[:, '수학']
```

## 10. 첫 번째 행의 첫 번째 값

```python
df.iloc[0, 0]
```

## 11. 김민준의 국어 점수

```python
df.loc[0, '국어']
```

## 12. 최예린의 영어 점수 - `iloc`

```python
df.iloc[3, 3]
```

## 13. 최예린의 영어 점수 - `loc`

```python
df.loc[3, '영어']
```

## 14. 처음 3개 행 선택 - `iloc`

```python
df.iloc[0:3]
```

## 15. 인덱스 0~2 행 선택 - `loc`

```python
df.loc[0:2]
```

## 16. `loc`과 `iloc` 슬라이싱의 차이

`iloc[0:2]`는 0, 1 위치만 포함하고 **끝 위치 2는 제외**합니다.

반면 `loc[0:2]`는 인덱스 라벨 0, 1, 2를 **모두 포함**합니다.

```python
df.iloc[0:2]
```

```python
df.loc[0:2]
```

## 17. 처음 3개 행에서 이름~수학 열 선택 - `iloc`

```python
df.iloc[0:3, 0:3]
```

## 18. 인덱스 0~2 행에서 이름~수학 열 선택 - `loc`

```python
df.loc[0:2, '이름':'수학']
```

## 19. 특정 행 여러 개 선택 - `iloc`

```python
df.iloc[[0, 2, 4]]
```

## 20. 특정 행 여러 개 선택 - `loc`

```python
df.loc[[0, 2, 4]]
```

## 21. 특정 열 여러 개 선택 - `iloc`

```python
df.iloc[:, [0, 4, 5]]
```

## 22. 특정 열 여러 개 선택 - `loc`

```python
df.loc[:, ['이름', '총점', '평균']]
```

## 23. 특정 행/열을 동시에 선택 - `iloc`

```python
df.iloc[[0, 2, 4], [0, 1, 5]]
```

## 24. 특정 행/열을 동시에 선택 - `loc`

```python
df.loc[[0, 2, 4], ['이름', '국어', '평균']]
```

## 25. 마지막 행 - `iloc`

```python
df.iloc[-1]
```

## 26. 뒤에서 두 번째 행 - `iloc`

```python
df.iloc[-2]
```

## 27. 마지막 열 - `iloc`

```python
df.iloc[:, -1]
```

## 28. 마지막 두 열 - `iloc`

```python
df.iloc[:, -2:]
```

## 29. 조건식과 `loc` 함께 사용

`loc`은 Boolean 조건식과 함께 매우 자주 사용됩니다.

```python
df.loc[df['평균'] >= 85]
```

```python
df.loc[df['평균'] >= 85, ['이름', '평균']]
```

## 30. 2학년 학생의 이름과 평균

```python
df.loc[df['학년'] == 2, ['이름', '평균']]
```

## 31. 국어 85점 이상 학생의 이름과 국어 점수

```python
df.loc[df['국어'] >= 85, ['이름', '국어']]
```

## 32. 수학 80점 이상이면서 영어 85점 이상

```python
df.loc[(df['수학'] >= 80) & (df['영어'] >= 85), ['이름', '수학', '영어']]
```

## 33. 1학년 또는 2학년 학생

```python
df.loc[df['학년'].isin([1, 2]), ['이름', '학년']]
```

## 34. 평균 80~85점 학생

```python
df.loc[df['평균'].between(80, 85), ['이름', '평균']]
```

# 값 수정 연습

`loc`과 `iloc`은 조회뿐 아니라 특정 셀이나 범위의 값을 수정하는 데도 사용할 수 있습니다.

아래 예제는 원본 `df`를 보호하기 위해 복사본 `df_edit`을 사용합니다.

```python
df_edit = df.copy()
df_edit
```

## 35. `loc`으로 김민준의 등급을 A로 변경

```python
df_edit.loc[0, '등급'] = 'A'
df_edit
```

## 36. `iloc`으로 두 번째 학생의 등급을 A로 변경

```python
df_edit.iloc[1, 7] = 'A'
df_edit
```

## 37. 평균 85점 이상 학생의 등급을 A로 변경

```python
df_edit.loc[df_edit['평균'] >= 85, '등급'] = 'A'
df_edit
```

## 38. 2학년 학생들의 등급을 A로 변경

```python
df_edit.loc[df_edit['학년'] == 2, '등급'] = 'A'
df_edit
```

# 인덱스를 이름으로 바꾼 뒤 `loc` 연습

`loc`의 장점을 더 잘 이해하기 위해 `이름`을 인덱스로 설정해 봅니다.

```python
df_name = df.set_index('이름')
df_name
```

## 39. 김민준 행 선택

```python
df_name.loc['김민준']
```

## 40. 박지호와 최예린 선택

```python
df_name.loc[['박지호', '최예린']]
```

## 41. 이서연의 수학 점수

```python
df_name.loc['이서연', '수학']
```

## 42. 김민준~박지호 범위 선택

```python
df_name.loc['김민준':'박지호']
```

## 43. 김민준~박지호의 국어~영어 점수

```python
df_name.loc['김민준':'박지호', '국어':'영어']
```
