## 4차시 · 공정 데이터 실습

배운 기능을 반도체 공정 데이터에 이어서 사용합니다.

### 준비 · 공정 이력 데이터

빈칸이 없는 셀입니다. `합격여부`는 1 = 합격, -1 = 불합격이고, 마지막 행은 중복입니다.

````python
process_df = pd.DataFrame({
    'lot_id': ['L001', 'L002', 'L003', 'L004', 'L005', 'L005'],
    '공정': ['식각', '증착', '식각', '세정', '증착', '증착'],
    '온도_섭씨': [298.5, 302.1, None, 299.8, 305.2, 305.2],
    '압력_Pa': [101.2, 99.8, 103.1, 100.5, 98.9, 98.9],
    '두께_nm': [100.4, 102.3, 98.7, 101.1, 104.8, 104.8],
    '합격여부': [1, 1, -1, 1, -1, -1]
})
print(process_df)
````

### Q26. 원본을 복사해서 정리하기

원본은 두고 복사본에서 온도 결측값을 중앙값으로 채우고 중복을 지웁니다.

<details>
<summary><strong>💡 힌트</strong></summary>

복사는 `copy()`, 중앙값은 `median()`입니다.

</details>

````python
work = process_df.___()
work['온도_섭씨'] = work['온도_섭씨'].fillna(work['온도_섭씨'].___())
work = work.drop_duplicates().reset_index(drop=True)
print(work)
````
