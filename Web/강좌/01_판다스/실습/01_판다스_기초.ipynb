### 라이브러리 불러오기
- 설명: 분석에 필요한 라이브러리를 불러오고 짧은 별칭을 정합니다.
- 수업 메모: 모든 코드의 시작 - import 문 작성하기

```python
# pandas 라이브러리를 불러옵니다
# 'as pd'는 앞으로 pandas를 pd라는 짧은 이름으로 부르겠다는 뜻입니다
import pandas as pd

print("✅ 라이브러리 불러오기 완료!")
print(f"Pandas 버전: {pd.__version__}")
```

### Series - 한 열의 데이터
- 설명: 인덱스 + 값의 쌍으로 이루어진 1차원 데이터 구조

```python
# 학생들의 수학 점수로 Series를 만들어봅니다
math_scores = pd.Series([85, 92, 78, 95, 88])

print("=== 수학 점수 Series ===")
print(math_scores)
```

```python
# 인덱스를 이름으로 지정할 수도 있습니다
math_scores_named = pd.Series(
    [85, 92, 78, 95, 88],
    index=['김민준', '이서연', '박지호', '최예린', '정도윤']
)

# 특정 학생 점수 조회
print(f"이서연 점수: {math_scores_named['이서연']}점")
```

```python
# 기본 통계 계산
print(f"평균: {math_scores_named.mean():.1f}점")
print(f"최고: {math_scores_named.max()}점")
```

### DataFrame - 표 형태의 데이터
- 설명: 여러 Series를 모아놓은 2차원 데이터 구조 (엑셀 시트)

```python
# 딕셔너리로 데이터를 만들고 DataFrame으로 변환합니다
# 딕셔너리의 '키' = 열(컬럼) 이름
# 딕셔너리의 '값' = 각 열의 데이터 목록

data = {
    '이름':  ['김민준', '이서연', '박지호', '최예린', '정도윤',
             '한지수', '오승현', '윤채원'],
    '학년':  [2, 1, 3, 2, 1, 3, 2, 1],
    '국어':  [85, 92, 78, 95, 88, 76, 91, 83],
    '수학':  [92, 78, 95, 88, 72, 85, 79, 96],
    '영어':  [88, 95, 82, 91, 85, 90, 77, 89],
    
}

# 딕셔너리를 DataFrame으로 변환!
df = pd.DataFrame(data)

print(df)  # Jupyter에서는 예쁜 표로 출력됩니다
```

### 데이터 기본 확인 함수들
- 설명: head, info, describe 등으로 데이터의 첫인상과 구조를 확인합니다.
- 수업 메모: 데이터를 처음 받으면 가장 먼저 해야 할 것들

```python
# head() - 앞에서 n줄만 보기 (기본: 5줄)
df.head(3)
```

```python
# tail() - 뒤에서 n줄 보기
df.tail(3)
```

```python
# shape - 데이터 크기 (행의 수, 열의 수)
print(f"크기: {df.shape}")  # (8, 6)
```

```python
# info() - 전체 정보 요약 (열 이름, 데이터 타입, 개수)
df.info()
```

```python
# describe() - 숫자 데이터의 통계 요약
# count, mean, std, min, 25%, 50%, 75%, max 를 한번에!
df.describe()
```

```python
# columns - 열 이름 목록
print(df.columns)
```

```python
# dtypes - 각 열의 데이터 타입
print(df.dtypes)
```

### 데이터 선택하기 (loc & iloc)
- 설명: 필요한 행과 열만 골라 보며 DataFrame 선택 문법을 연습합니다.
- 수업 메모: 원하는 행/열만 골라내는 방법

```python
## 열(컬럼) 선택 ##
# 열 하나 선택 → Series 반환
df['수학']
```

```python
# 여러 열 선택 → DataFrame 반환 (대괄호 두 겹!)
df[['이름', '수학', '영어']]
```

```python
## 행 선택 ##
# iloc - 위치(번호)로 선택 (0부터 시작!)
df.iloc[0]          # 첫 번째 행
```

```python
df.iloc[1:4]       # 1~3번 행 (4번 미포함)
```

```python
df.iloc[0:3, 0:3]  # 0~2행, 0~2열
```

```python
# loc - 인덱스 이름으로 선택
df.loc[0]              # 인덱스가 0인 행
```

```python
df.loc[0:2, ['이름', '수학']]  # loc은 끝 번호 포함!
```

## 마무리

- `pd.Series()`, `pd.DataFrame()`으로 데이터를 만들었습니다.
- `head()`, `info()`, `describe()`로 데이터의 첫인상을 확인했습니다.
- `loc`(이름)과 `iloc`(번호)으로 원하는 행·열을 골랐습니다.
