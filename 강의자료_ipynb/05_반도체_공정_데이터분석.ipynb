## 3단계 · 결측값 & 중복 처리

### 2 결측값 찾기
- 원본 코드: `07_check_missing.py`
- 설명: 결측값을 찾고 삭제 또는 대체 방식으로 정리합니다.

````python
# ① 열별 결측값 개수 (True를 1로 세어줌)
print(df.isnull().sum())

# ② 전체 결측값 개수 (모든 열을 합산)
print(f"전체 결측값 개수: {df.isnull().sum().sum()}개")
````

### 3 결측값 처리하는 3가지 방법
- 원본 코드: `08_fill_missing.py`
- 설명: 결측값을 찾고 삭제 또는 대체 방식으로 정리합니다.

````python
# 원본 보호를 위해 복사본을 만들어 작업
df_filled = df.copy()

# 숫자 컬럼만 골라서 평균으로 채우기
numeric_cols = ['온도_섭씨', '압력_Pa', '가스유량_slm', '전력_W',
                '진공도_mTorr', '두께_nm', '습도_pct', '진동_mm_s',
                '처리시간_sec', '냉각수온도_섭씨']

for col in numeric_cols:
    mean_val = df_filled[col].mean()
    df_filled[col] = df_filled[col].fillna(mean_val)

# 처리 후 결측 확인
print(f"처리 후 남은 결측값: {df_filled.isnull().sum().sum()}개")
````

### 4 중복된 행 찾고 제거하기
- 원본 코드: `09_duplicates.py`
- 설명: 중복 행을 확인하고 제거한 뒤 인덱스를 정리합니다.

````python
# ① 중복된 행 개수 확인
print(f"중복 행 수: {df_filled.duplicated().sum()}개")

# ② 중복 제거
df_clean = df_filled.drop_duplicates()

# ③ 인덱스 다시 매기기 (중복 제거 후 번호가 띄엄띄엄 됨)
df_clean = df_clean.reset_index(drop=True)

print(f"정리 전: {len(df_filled)}행 → 정리 후: {len(df_clean)}행")
````

## 4단계 · 탐색적 데이터 분석 (EDA)

### 1 분석에 필요한 새 컬럼 만들기
- 원본 코드: `10_new_columns.py`
- 설명: 1 분석에 필요한 새 컬럼 만들기 단계에서 사용하는 핵심 코드를 실행해 봅니다.

````python
# df_clean을 앞으로 계속 사용 (결측·중복 처리된 깨끗한 표)

# 합격여부 1/-1 → '합격'/'불합격' 문자열로 바꿔 보기 좋게
df_clean['result'] = df_clean['합격여부'].map({1: '합격', -1: '불합격'})

# 결과 확인
df_clean[['온도_섭씨', '두께_nm', '합격여부', 'result']].head()
````

### 2 조건에 맞는 행만 골라내기
- 원본 코드: `11_filter.py`
- 설명: 조건식을 사용해 관심 있는 행만 추출합니다.

````python
# ① 불합격 데이터만 골라내기
fail_df = df_clean[df_clean['합격여부'] == -1]
print(f"불합격 건수: {len(fail_df)}건")

# ② 온도가 305도를 넘는 공정
high_temp_df = df_clean[df_clean['온도_섭씨'] > 305]
print(f"고온 공정 수: {len(high_temp_df)}건")

# ③ 여러 조건 동시에 (AND는 &, OR은 |, 괄호 필수!)
condition = (df_clean['온도_섭씨'] > 305) & (df_clean['두께_nm'] > 103)
risk_df = df_clean[condition]
print(f"고온+두꺼움 공정: {len(risk_df)}건")
````

### 3 groupby — 그룹별로 통계 내기
- 원본 코드: `12_groupby.py`
- 설명: 그룹별 평균과 통계를 구해 조건별 차이를 비교합니다.

````python
# 숫자 컬럼 리스트 (3단계에서 정의한 것과 동일)
numeric_cols = ['온도_섭씨', '압력_Pa', '가스유량_slm', '전력_W',
                '진공도_mTorr', '두께_nm', '습도_pct', '진동_mm_s',
                '처리시간_sec', '냉각수온도_섭씨']

# 합격여부별 모든 컬럼의 평균
group_mean = df_clean.groupby('result')[numeric_cols].mean().round(2)
print(group_mean.T)   # .T 는 행·열 뒤집기 (보기 편하게)
````

### 5 히스토그램 — 값이 어떻게 퍼져 있는지 보기

### 6 박스플롯 — 합격 vs 불합격 한눈에 비교

### 7 상관관계 히트맵 — 변수들 간의 관계

### 실습 3 · IQR로 이상값 후보 찾기

**목표:** 온도 분포에서 일반 범위를 크게 벗어난 측정값을 점검 대상으로 표시합니다.

````python
# TODO 1: 온도_섭씨의 Q1(25%)과 Q3(75%)를 구하세요.
# TODO 2: IQR = Q3 - Q1을 계산하세요.
# TODO 3: Q1 - 1.5*IQR 미만 또는 Q3 + 1.5*IQR 초과 행을 찾으세요.
# 주의: 이상값 후보가 반드시 오류나 불량이라는 뜻은 아닙니다.
````

<details>
<summary><strong>힌트 보기</strong></summary>

`quantile(0.25)`와 `quantile(0.75)`를 사용하고 두 조건은 `|`로 연결합니다.

</details>

**정답 예제:** 먼저 직접 시도한 뒤 아래 셀을 실행하세요.


````python
q1 = practice_clean['온도_섭씨'].quantile(0.25)
q3 = practice_clean['온도_섭씨'].quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

temperature_outliers = practice_clean[
    (practice_clean['온도_섭씨'] < lower) |
    (practice_clean['온도_섭씨'] > upper)
]

print(f'온도 일반 범위: {lower:.2f} ~ {upper:.2f} °C')
print(f'이상값 후보: {len(temperature_outliers)}건')
temperature_outliers[['온도_섭씨', '압력_Pa', '두께_nm', '판정']].head()
````
