## 2차시: 탐색적 데이터 분석 (EDA)

### 결측값 확인 및 처리
- 원본 코드: `01_missing_values.py`
- 설명: 결측값을 찾고 삭제 또는 대체 방식으로 정리합니다.
- 수업 메모: NaN(Not a Number) - 값이 없는 빈 칸 다루기

````python
import pandas as pd


# 결측값이 포함된 데이터 만들기 (np.nan = 값 없음)
data = {
    '이름':  ['김민준', '이서연', '박지호', '최예린', '정도윤'],
    '국어':  [85, 92, 78, None, 88],   # 최예린 국어 결측
    '수학':  [92, 78, None, 88, 72],   # 박지호 수학 결측
    '영어':  [88, None, 82, 91, 85]    # 이서연 영어 결측
}
df = pd.DataFrame(data)

## 결측값 확인 ##
# isnull() → 각 값이 결측인지 True/False
df.isnull()

# 열별 결측값 개수
print(df.isnull().sum())

# 결측값 비율 계산
missing_ratio = (df.isnull().sum() / len(df) * 100).round(1)
print(missing_ratio)

## 결측값 처리 ##

# 방법 1: 결측값 있는 행 삭제
df_dropped = df.dropna()

# 방법 2: 평균으로 채우기
df_filled = df.copy()  # 원본 보존을 위해 복사
for col in ['국어', '수학', '영어']:
    col_mean = df_filled[col].mean()
    df_filled[col] = df_filled[col].fillna(col_mean).round(1)

# 방법 3: 특정 값으로 채우기
df.fillna(0)        # 0으로 채우기
df.fillna('없음')   # 문자로 채우기
````

### 중복 데이터 처리
- 원본 코드: `02_duplicates.py`
- 설명: 중복 행을 확인하고 제거한 뒤 인덱스를 정리합니다.
- 수업 메모: 똑같은 행이 두 번 이상 있는 경우 찾아서 제거하기

````python
# duplicated() - 중복 여부 확인 (True = 중복)
print(df.duplicated())

# 중복된 행이 몇 개인지
print(f"중복 행 수: {df.duplicated().sum()}개")

# 실제로 어떤 행이 중복인지 확인
df[df.duplicated()]

# drop_duplicates() - 중복 제거!
df_clean = df.drop_duplicates()

# 인덱스를 0부터 새로 정렬 (중복 제거 후 인덱스가 비어있을 수 있어서)
df_clean = df_clean.reset_index(drop=True)
````

### 새 열 추가 & 조건 필터링
- 원본 코드: `03_filter.py`
- 설명: 조건식을 사용해 관심 있는 행만 추출합니다.
- 수업 메모: 계산된 열 추가하고 조건에 맞는 행만 선택하기

````python
## 새 열 추가 ##
df['총점'] = df['국어'] + df['수학'] + df['영어'] 
df['평균'] = (df['총점'] / 3).round(1)
df['학년'] = [1, 1, 2, 2, 3]


# apply()로 함수를 각 값에 적용 (등급 부여)
def get_grade(score):
    if   score >= 90: return 'A'
    elif score >= 80: return 'B'
    elif score >= 70: return 'C'
    else:             return 'D'

df['등급'] = df['평균'].apply(get_grade)

## 조건 필터링 ##
# 평균 85점 이상인 학생만
high = df[df['평균'] >= 85]

# 여러 조건 조합 (& = AND, | = OR)
cond = (df['학년'] == 1) & (df['평균'] >= 80)
result = df[cond]

# 특정 값 포함 여부: isin()
top = df[df['등급'].isin(['A', 'B'])]

## 정렬 ##
# 평균 내림차순 정렬
df.sort_values('평균', ascending=False)
````

### groupby - 그룹별 집계
- 원본 코드: `04_groupby.py`
- 설명: 그룹별 평균과 통계를 구해 조건별 차이를 비교합니다.
- 수업 메모: 엑셀의 피벗 테이블과 같은 강력한 집계 기능

````python
# 학년별 평균 점수
grade_avg = df.groupby('학년')[['국어', '수학', '영어']].mean().round(1)

# 학년별 학생 수
student_count = df.groupby('학년').size()

# 여러 통계 한번에: agg()
grade_stats = df.groupby('등급')['평균'].agg(['count', 'mean', 'min', 'max'])
grade_stats.columns = ['인원수', '평균', '최저', '최고']

print(grade_stats.round(1))
````

### 박스 플롯 — plt.boxplot()
- 원본 코드: `05_박스플롯_색추가.py`
- 설명: 박스플롯으로 그룹별 분포와 이상치를 비교합니다.
- 수업 메모: 데이터가 어떻게 퍼져 있는지 한 눈에 볼 때 사용합니다

````python
# patch_artist=True → 상자 안을 색으로 채우기
bp = plt.boxplot([국어, 수학, 영어],
                  labels=['국어', '수학', '영어'],
                  patch_artist=True)

# 각 상자에 색 지정
색목록 = ['lightblue', 'lightyellow', 'lightgreen']
for 상자, 색 in zip(bp['boxes'], 색목록):
    상자.set_facecolor(색)

plt.title('과목별 점수 분포')
plt.ylabel('점수')
plt.show()
````
